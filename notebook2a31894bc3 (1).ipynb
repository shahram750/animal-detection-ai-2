```python

import tensorflow as tf
from pathlib import Path
from PIL import Image

# Original dataset path
source_path = Path(
    "/kaggle/input/datasets/shahram06/"
    "animal-detction/Animal detection ai"
)

# Converted images will be stored here
output_path = Path("/kaggle/working/animal_dataset")
output_path.mkdir(parents=True, exist_ok=True)

classes = ["cats", "lions", "parrots", "sparrows", "tigers"]

extensions = {
    ".jpg", ".jpeg", ".jfif", ".png",
    ".webp", ".bmp", ".gif"
}

# Convert every image to JPG
for class_name in classes:
    source_folder = source_path / class_name
    target_folder = output_path / class_name
    target_folder.mkdir(parents=True, exist_ok=True)

    count = 0

    for file in source_folder.rglob("*"):
        if not file.is_file():
            continue

        if file.suffix.lower() not in extensions:
            continue

        try:
            with Image.open(file) as img:
                img.convert("RGB").save(
                    target_folder / f"image_{count}.jpg",
                    format="JPEG"
                )
            count += 1

        except Exception as e:
            print("Skipped:", file.name, "|", e)

    print(class_name, ":", count, "images converted")

print("\nConverted dataset:", output_path)
```

```python

image_size = (160, 160)
batch_size = 4
seed = 42

train_ds = tf.keras.utils.image_dataset_from_directory(
    str(output_path),
    validation_split=0.2,
    subset="training",
    seed=seed,
    image_size=image_size,
    batch_size=batch_size,
    label_mode="int",
    class_names=classes
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    str(output_path),
    validation_split=0.2,
    subset="validation",
    seed=seed,
    image_size=image_size,
    batch_size=batch_size,
    label_mode="int",
    class_names=classes
)

class_names = train_ds.class_names

print("Classes:", class_names)
print("Training batches:", tf.data.experimental.cardinality(train_ds).numpy())
print("Validation batches:", tf.data.experimental.cardinality(val_ds).numpy())
```

```python

import tensorflow as tf
from pathlib import Path

dataset_path = Path("/kaggle/working/animal_dataset")

image_size = (160, 160)
batch_size = 2

train_ds = tf.keras.utils.image_dataset_from_directory(
    str(dataset_path),
    image_size=image_size,
    batch_size=batch_size,
    label_mode="int",
    shuffle=True,
    seed=42
)

class_names = train_ds.class_names

print("Classes:", class_names)
print("Number of classes:", len(class_names))
print("Images:", sum(
    len(list((dataset_path / c).glob("*.jpg")))
    for c in class_names
))

train_ds = train_ds.prefetch(tf.data.AUTOTUNE)
```

```python

from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.applications import MobileNetV2

# Pretrained model
base_model = MobileNetV2(
    input_shape=(160, 160, 3),
    include_top=False,
    weights="imagenet"
)

base_model.trainable = False

# Build our animal classifier
inputs = keras.Input(shape=(160, 160, 3))

x = layers.Rescaling(
    scale=1.0 / 127.5,
    offset=-1
)(inputs)

x = base_model(x, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.2)(x)

outputs = layers.Dense(
    len(class_names),
    activation="softmax"
)(x)

model = keras.Model(inputs, outputs)

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# Train the model for a small test run
history = model.fit(
    train_ds,
    epochs=5
)
```

Training aur loss graph****

```python

import matplotlib.pyplot as plt

# Training accuracy and loss graphs
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(history.history["accuracy"], marker="o")
plt.title("Training Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(history.history["loss"], marker="o")
plt.title("Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python

import numpy as np
import ipywidgets as widgets

from PIL import Image
from IPython.display import display

upload = widgets.FileUpload(
    accept="image/*",
    multiple=False,
    description="Upload Animal Photo"
)

display(upload)

def predict_animal(change):
    if not upload.value:
        return

    uploaded = upload.value

    # Support different ipywidgets versions
    if isinstance(uploaded, dict):
        file_info = next(iter(uploaded.values()))
    else:
        file_info = uploaded[0]

    image_bytes = bytes(file_info["content"])

    image = Image.open(
        __import__("io").BytesIO(image_bytes)
    ).convert("RGB")

    display(image)

    image = image.resize((160, 160))
    image_array = np.array(image, dtype=np.float32)
    image_array = np.expand_dims(image_array, axis=0)

    predictions = model.predict(image_array, verbose=0)[0]

    best_index = int(np.argmax(predictions))
    animal_name = class_names[best_index]
    confidence = float(predictions[best_index]) * 100

    print("Predicted Animal:", animal_name.upper())
    print(f"Model Score: {confidence:.2f}%")

    print("\nAll class scores:")
    for i, name in enumerate(class_names):
        print(f"{name}: {predictions[i] * 100:.2f}%")

upload.observe(predict_animal, names="value")
```
